```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import numpy as np
import plotly.graph_objects as go
```

```python
df = pd.read_csv(r"C:\Users\sdjkl\CG\VRBioTalk\results.csv")
df.head()
```

```python
df["average_energy"] = np.sqrt(
    df["average_flux_x"]**2 +
    df["average_flux_y"]**2 +
    df["average_flux_z"]**2
)
```

```python
pivot = df.pivot(index="bounces", columns="samples", values="average_energy")

x = pivot.columns.to_numpy()
y = pivot.index.to_numpy()
Z = pivot.to_numpy()

fig = go.Figure(
    data=[
        go.Surface(
            x=x,
            y=y,
            z=Z,
            colorbar=dict(title="average_energy")
        )
    ]
)

fig.update_layout(
    title="Height Field of average_energy",
    scene=dict(
        xaxis_title="samples",
        yaxis_title="bounces",
        zaxis_title="average_energy",
    ),
    width=900,
    height=700,
)

fig.show()
```
